```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

#### Step 1: Load the dataset


```python
file_path = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv"

df = pd.read_csv(file_path)
```

```python
pd.set_option('display.max_columns', None)
df.head()
```


### Task 1: Exploring Relationships with Scatter Plots



#### 1. Scatter Plot for Age vs. Job Satisfaction



Visualize the relationship between respondents' age (`Age`) and job satisfaction (`JobSatPoints_6`). Use this plot to identify any patterns or trends.




```python
## Write your code here
df_age_sat = df[['Age', 'JobSatPoints_6']].dropna()
df_age_sat
```

### Task 2: Enhancing Scatter Plots


#### 1. Scatter Plot with Trend Line for Age vs. Job Satisfaction



Add a regression line to the scatter plot of Age vs. JobSatPoints_6 to highlight trends in the data.

```python
sns.regplot(
    data=df_age_sat,
    x='Age_numeric',
    y='JobSatPoints_6'
)

plt.title('Correlation between Age and Job Satisfaction Points')
plt.xlabel('Age (Appriximate)')
plt.ylabel('Job Satisfaction Points (6)')
```

Weak negative correlation between Age and Job Satisfaction

Visualize the relationship between Age (`Age`) and Work Experience (`YearsCodePro`) using a scatter plot.

### Task 3: Combining Scatter Plots with Additional Features


#### 1. Bubble Plot of Compensation vs. Job Satisfaction with Age as Bubble Size



Create a bubble plot to explore the relationship between yearly compensation (`ConvertedCompYearly`) and job satisfaction (`JobSatPoints_6`), with bubble size representing age.


```python
## Write your code here
df_bubble = df[['ConvertedCompYearly', 'JobSatPoints_6', 'Age']].dropna()
df_bubble['Age_num'] = df_bubble['Age'].map(age_map)
df_bubble = df_bubble.dropna(subset=['Age_num'])
```

```python
sns.scatterplot(
    data=df_bubble,
    x='ConvertedCompYearly',
    y='JobSatPoints_6',
    size='Age_num',
    sizes=(20, 400),
    alpha=0.6
)

plt.xscale('log') 
plt.title('Bubble Plot: Compensation vs Job Satisfaction (Bubble Size = Age)')
plt.xlabel('Yearly Compensation (log scale)')
plt.ylabel('Job Satisfaction Points (6)')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
```

The bubble plot shows a positive but dispersed relationship between yearly compensation and job satisfaction. Larger bubbles (older respondents) tend to cluster at higher compensation levels, reflecting increased earnings with experience. However, job satisfaction varies widely across all compensation and age levels, indicating that higher pay does not necessarily guarantee higher job satisfaction.

#### 2. Scatter Plot for Popular Programming Languages by Job Satisfaction


Visualize the popularity of programming languages (`LanguageHaveWorkedWith`) against job satisfaction using a scatter plot. Use points to represent satisfaction levels for each language.


```python
df_lang_sat = df[['LanguageHaveWorkedWith', 'JobSatPoints_6']].dropna()
df_lang_sat
```

```python
df_lang_sat['Language'] = df_lang_sat['LanguageHaveWorkedWith'].str.split(';')
df_lang_sat = df_lang_sat.explode('Language')
df_lang_sat
```

```python
# Aggregate job satisfaction per language

lang_stats = (
    df_lang_sat
    .groupby('Language')
    .agg(
        AvgJobSat=('JobSatPoints_6', 'mean'),
        Count=('JobSatPoints_6', 'size')
    )
    .reset_index()
)
lang_stats.head()
```

```python
sns.scatterplot(
    data=lang_stats,
    x='Count',
    y='AvgJobSat',
    size='Count',
    sizes=(40, 400),
    alpha=0.7
)

# plt.xscale('log')
plt.title('Programming Language Popularity vs Job Satisfaction')
plt.xlabel('Number of Respondents')
plt.ylabel('Average Job Satisfaction (6)')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
```

The scatter plot shows that highly popular programming languages tend to cluster around moderate job satisfaction levels, while less common languages sometimes exhibit higher average satisfaction. This suggests that niche or specialized languages may be associated with more fulfilling roles, whereas widely used languages span a broader range of satisfaction outcomes

### Task 4: Scatter Plot Comparisons Across Groups


#### 1. Scatter Plot for Compensation vs. Job Satisfaction by Employment Type


Visualize the relationship between yearly compensation (`ConvertedCompYearly`) and job satisfaction (`JobSatPoints_6`), categorized by employment type (`Employment`). Use color coding or markers to differentiate between employment types.


```python
df['Employment'].unique()
```

```python
## Write your code here
df_emp_sat = df[['ConvertedCompYearly', 'JobSatPoints_6', 'Employment']].dropna()
```

```python
df_emp_sat['Employment_main'] = df_emp_sat['Employment'].str.split(';').str[0]
df_emp_sat
```

```python
sns.scatterplot(
    data=df_emp_sat,
    x='ConvertedCompYearly',
    y='JobSatPoints_6',
    hue='Employment_main',
    alpha=0.6
)

plt.xscale('log')
plt.title('Compensation vs Job Satisfaction by Employment Type')
plt.xlabel('Yearly Compensation (log scale)')
plt.ylabel('Job Satisfaction Points (6)')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
```

The scatter plot reveals noticeable differences in compensation and job satisfaction across employment types. Full-time employees tend to cluster at higher compensation levels, while freelancers and contractors display wider variability in both pay and satisfaction. Students and part-time workers generally appear at lower compensation levels with more dispersed satisfaction scores.

#### 2. Scatter Plot for Work Experience vs. Age Group by Country


Compare work experience (`YearsCodePro`) across different age groups (`Age`) and countries (`Country`). Use colors to represent different countries and markers for age groups.


```python
## Write your code here
df_exp = df[['YearsCodePro', 'Age', 'Country']].dropna()

df_exp['YearsCodePro'] = (
    df_exp['YearsCodePro']
    .replace({'Less than 1 year': 0.5, 'More than 50 years': 51})
    .astype(float)
)
```

```python
df_exp['Age_num'] = df_exp['Age'].map(age_map)
df_exp = df_exp.dropna(subset=['Age_num'])
```

```python
top_countries = df_exp['Country'].value_counts().head(5).index
df_exp = df_exp[df_exp['Country'].isin(top_countries)]
```

```python
sns.scatterplot(
    data=df_exp,
    x='YearsCodePro',
    y='Age_num',
    hue='Country',
    style='Age',
    alpha=0.6
)

plt.title('Professional Coding Experience Across Age Groups and Countries')
plt.xlabel('Years of Professional Coding Experience')
plt.ylabel('Age (Approximate)')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
```

The scatter plot shows a clear positive relationship between age and professional coding experience across all countries. Respondents from different countries follow similar experience-age trajectories, though some countries exhibit wider spreads, suggesting earlier entry into professional coding or longer career longevity. Younger age groups cluster at lower experience levels, while older groups show greater dispersion in years of professional coding.
