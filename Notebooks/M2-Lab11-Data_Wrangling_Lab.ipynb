### 4. Encoding Categorical Variables


<h5>4.1 Encode the Employment column using one-hot encoding.</h5>


```python
## Write your code here
employment_encoded = pd.get_dummies(df['Employment'], prefix='Employment')

df = pd.concat([df, employment_encoded], axis=1)
df
```

### 5. Handling Missing Values


<h5>5.1 Identify columns with the highest number of missing values.</h5>


```python
## Write your code here
df.isna().sum().sort_values(ascending=False)
```

<h5>5.2 Impute missing values in numerical columns (e.g., `ConvertedCompYearly`) with the mean or median.</h5>


```python
## Write your code here
df_numeric = df.select_dtypes('number')
df_numeric.columns
```

```python
numeric_columns = ['ResponseId', 'CompTotal', 'WorkExp', 'JobSatPoints_1',
       'JobSatPoints_4', 'JobSatPoints_5', 'JobSatPoints_6', 'JobSatPoints_7',
       'JobSatPoints_8', 'JobSatPoints_9', 'JobSatPoints_10',
       'JobSatPoints_11', 'ConvertedCompYearly', 'JobSat']

for col in numeric_columns:
    median_val = df[col].median()
    df[col] = df[col].fillna(median_val)

df[numeric_columns]
```

<h5>5.3 Impute missing values in categorical columns (e.g., `RemoteWork`) with the most frequent value.</h5>


```python
df['RemoteWork'].isna().sum()
```

```python
## Write your code here
most_frequent = df['RemoteWork'].mode()[0]
most_frequent
```

```python
df['RemoteWork'] = df['RemoteWork'].fillna(most_frequent)
df['RemoteWork'].isna().sum()
```

### 6. Feature Scaling and Transformation


<h5>6.1 Apply Min-Max Scaling to normalize the `ConvertedCompYearly` column.</h5>


```python
## Write your code here
min_val = df['ConvertedCompYearly'].min()
max_val = df['ConvertedCompYearly'].max()

df['ConvertedCompYearly'] = (df['ConvertedCompYearly'] - min_val) / (max_val - min_val)
df['ConvertedCompYearly']
```

<h5>6.2 Log-transform the ConvertedCompYearly column to reduce skewness.</h5>


```python
import numpy as np
```

```python
## Write your code here
df['ConvertedCompYearly'] = np.log1p(df['ConvertedCompYearly'])
df['ConvertedCompYearly']
```

### 7. Feature Engineering


<h5>7.1 Create a new column `ExperienceLevel` based on the `YearsCodePro` column:</h5>


```python
## Write your code here
df['YearsCodePro'] = pd.to_numeric(df['YearsCodePro'], errors='coerce')
df['YearsCodePro']
```

```python
df['YearsCodePro'] = df['YearsCodePro'].fillna(df['YearsCodePro'].median())
df['YearsCodePro']
```

```python
def categiorise_years(years):
    if years < 2:
        return 'Junior'
    elif 2 <= years < 5:
        return 'Mid-Level'
    elif 5 <= years < 10:
        return 'Senior'
    else:
        return 'Expert'
    
df['ExperienceLevel'] = df['YearsCodePro'].apply(categiorise_years)
df['ExperienceLevel']
```
