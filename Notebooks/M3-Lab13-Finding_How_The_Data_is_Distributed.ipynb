### Step 3: Handle Missing Data


- Identify missing values in the dataset.

- Impute or remove missing values as necessary to ensure data completeness.



```python
## Write your code here
df.isna().sum()
```

```python
pd.set_option('display.max_columns', None)
df.select_dtypes('object').head()
```

```python
df['YearsCodePro'] = pd.to_numeric(df['YearsCodePro'], errors='coerce')

df['YearsCodePro'] = df['YearsCodePro'].fillna(df['YearsCodePro'].mean())

df['YearsCodePro']
```

```python
number_cols = df.select_dtypes('number').columns

# Excluding the first numeric column
number_cols = number_cols[1:]

df[number_cols] = df[number_cols].fillna(
    df[number_cols].mean().iloc[0]
)
```

```python
object_cols = df.select_dtypes(include='object').columns

df[object_cols] = df[object_cols].fillna(
    df[object_cols].mode().iloc[0]
)
df.head()
```

### Step 5: Visualize Job Satisfaction (Focus on JobSat)


- Create a pie chart or KDE plot to visualize the distribution of `JobSat`.

- Provide an interpretation of the plot, highlighting key trends in job satisfaction.


```python
## Write your code here
job_counts = df['JobSat'].value_counts().head(9)

plt.figure(figsize=(6, 6))
plt.pie(job_counts, labels=job_counts.index, autopct='%1.1f%%')
plt.title('Distribution of Job Satisfaction')
plt.show()

# The distribution indicates high job satisfaction among most respondents.
```

### Step 6: Programming Languages Analysis


- Compare the frequency of programming languages in `LanguageHaveWorkedWith` and `LanguageWantToWorkWith`.
  
- Visualize the overlap or differences using a Venn diagram or a grouped bar chart.


```python
## Write your code here
have = df['LanguageHaveWorkedWith'].str.split(';').explode()
want = df['LanguageWantToWorkWith'].str.split(';').explode()

have_counts = have.value_counts()
want_counts = want.value_counts()

have_set = set(have.dropna().unique())
want_set = set(want.dropna().unique())

counts_df = pd.DataFrame({
    'Have Worked With': have_counts,
    'Want to Work With': want_counts
}).fillna(0)
```

```python
counts_df.head().plot(kind='bar', figsize=(12,6))
plt.title('Programming Languages: Worked With vs Want to Work With')
plt.xlabel('Programming Language')
plt.ylabel('Number of Respondents')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Step 7: Analyze Remote Work Trends


- Visualize the distribution of RemoteWork by region using a grouped bar chart or heatmap.


```python
## Write your code here
# Group the data
remote_country = df.groupby(["Country", "RemoteWork"]).size().reset_index(name="Count")

pivot_df = remote_country.pivot(index="Country", columns="RemoteWork", values="Count").fillna(0)

plt.figure(figsize=(12, 8))
sns.heatmap(pivot_df, cmap="viridis")
plt.title("Heatmap: Remote Work Distribution by Country")
plt.xlabel("Remote Work Type")
plt.ylabel("Country")
plt.show()
```

### Step 8: Correlation between Job Satisfaction and Experience


- Analyze the correlation between overall job satisfaction (`JobSat`) and `YearsCodePro`.
  
- Calculate the Pearson or Spearman correlation coefficient.

```python
## Write your code here
pearson_corr = df['JobSat'].corr(df['YearsCodePro'], method='pearson')
print("Pearson correlation:", pearson_corr)
```
